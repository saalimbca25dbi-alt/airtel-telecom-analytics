```python
import sqlite3
import pandas as pd

con = sqlite3.connect(r'C:\Users\saali\Desktop\airtel-telecom-analytics\data\raw\airtel.db') 

def q(sql):
    return pd.read_sql(sql, con)
```

```python
q(""" 
select connection_type, count(*) as subs,
       round(100.0*count (*)/22000,2) as pct 
from customer_features
group by connection_type order by subs desc;
""")
```
